# Quick start

Install and sync the environment:

```bash
uv sync
```

To open the interactive launcher (Play, Editor, Settings):

```bash
python -m factoriax
```

## Programmatic usage

Create an environment, reset it, and take a step:

````python
import jax
import factoriax

env, params = factoriax.make("EasyRocket-v1")

key = jax.random.PRNGKey(0)
key, reset_key, step_key, action_key = jax.random.split(key, 4)

obs, state = env.reset_env(reset_key, params)

action = jax.random.randint(action_key, shape=(), minval=0, maxval=len(factoriax.Action))
obs, state, reward, done, info = env.step_env(step_key, state, action, params)

print("obs shape:", obs.shape)
print("reward:   ", reward)
````
